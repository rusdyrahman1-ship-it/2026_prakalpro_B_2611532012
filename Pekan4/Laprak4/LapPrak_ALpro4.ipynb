## 2.1 Dasar Teori
**1. Pengenalan Struktur Percabangan Python**

Struktur percabangan dalam Python adalah suatu struktur yang digunakan untuk membuat program dapat mengambil keputusan dan menjalankan blok kode tertentu berdasarkan suatu kondisi, seperti kondisi benar (True) atau salah (False). Python menyediakan berbagai jenis struktur percabangan, di antaranya percabangan if, percabangan if-else, percabangan if-elif-else, dan percabangan match-case. Dengan memahami struktur percabangan, seorang programmer dapat menuliskan alur logika program yang lebih dinamis, efisien, dan mudah dipahami.

**2. Konsep Struktur Percabangan Python**

Struktur percabangan merupakan bagian dari program yang digunakan untuk melakukan pengambilan keputusan terhadap suatu kondisi tertentu. Di dalam bahasa pemrograman, seperti Python, struktur percabangan digunakan untuk mengatur alur eksekusi program agar dapat menjalankan blok kode yang berbeda-beda, misalnya untuk memeriksa suatu syarat, membandingkan nilai, atau memilih salah satu dari beberapa kemungkinan tindakan. Setiap struktur percabangan memiliki kata kunci tertentu (seperti if, elif, else, atau match-case) yang membedakannya dari struktur percabangan lain. Jika kondisi yang digunakan salah atau tidak sesuai dengan logika yang diharapkan, maka struktur percabangan tersebut dapat menghasilkan alur program yang tidak berjalan sebagaimana mestinya. Dengan menggunakan struktur percabangan, sebuah program dapat memproses, membandingkan, dan menentukan tindakan yang sesuai berdasarkan nilai yang tersimpan di dalam variabel.

**3. Jenis Struktur Percabangan dalam Python**

Ada 4 jenis struktur percabangan dalam Python, yaitu:
1. Struktur percabangan if. Percabangan if biasanya digunakan untuk menangkap kasus untuk membuat sebuah pernyataan kondisional. Jika hasilnya True maka blok kode di bawah if akan dijalankan dan jika hasilnya false maka blok kode di bawah if akan dilewatkan. 
2. Struktur percabangan if - else. Percabangan if - else biasanya digunakan untuk menentukan tindakan berdasarkan suatu kondisi tertentu, melainkan digunakan untuk menentukan tindakan jika kondisi yang dinyatakan tidak sesuai. percabangan if digunakan untuk mengeksekusi kondisi True dan percabangan else digunakan untuk mengeksekusi kondisi False.
3. Struktur percabangan if - elif - else. Percabangan if - elif - else digunakan apabila lebih dari dua pilihan keputusan / kondisi. Yang mana percabangan elif digunakan untuk memeriksa kondisi tambahan jika kondisi if di atasnya bernilai salah.
4. Struktur percabangan match - case. Percabangan match - case merupakan bentuk sederhana dari percabangan if - elif - else. Match digunakan untuk memilih salah satu dari banyak blok kode yang akan dieksekusi. nilai dari match akan dibandingkan dengan nilai masing masing case. Apabila ada kecocokan antara nilai match dan nilai case, maka blok kode di bawah case akan dieksekusi.
