## 2.2 Kode Program

### 2.2.1 Program "NamaKelas_NIM"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 pratikum alpro
"""
print ("Hello World")
print()
print("program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + "dibuat oleh a\n"
    +  "Muhammad Rusdy Rahman dan 2611532012")
```

**Penjelasan :**
1. Bagian pertama, bagian ini adalah teks yang diapit tiga tanda kutip. Karena tidak disimpan ke variabel atau dipanggil, Python menganggapnya sebagai komentar dan tidak ikut dieksekusi/dicetak. Fungsinya hanya sebagai catatan/dokumentasi program (judul tugas, keterangan pertemuan, dll).
2. Bagian kedua,bagian ini berisi deretan print sederhana.Pada baris pertama, print() mencetak teks Hello World. Pada baris kedua, kode print mencetak baris kosong.Pada baris ketiga, print() mencetak program ini menghasilkan.Pada baris keempat, print() mencetak teks Empat baris output.Pada baris terakhir, berisi karakter escape \\, yang di dalam string Python setiap \\ mewakili satu karakter \ sehingga hasil cetaknya adalah: / \\ // \\\\ /// \\\\\\
3. Bagian ketiga, bagian ini berisi print() dengan dengan penggabungan string (concatenation) dan \n.Tanda + menggabungkan tiga potongan string jadi satu kalimat utuh.\n di tengah string berarti "pindah ke baris baru" saat dicetak.Kode ditulis menjadi dua baris fisik hanya untuk kerapian, Python tetap membacanya sebagai satu perintah print() karena tanda kurungnya belum ditutup di baris pertama.
