**2. Konsep Variabel dalam Python**

Variabel adalah tempat yang digunakan untuk menyimpan nilai.Di dalam bahasa pemrograman, seperti Python, variabel digunakan untuk menyimpan data yang dinginkan oleh program.Setiap variabel memiliki nama yang unik yang mengidentifikasinya.Jika ada 1 huruf yang berbeda dalam pengetikkan, maka variabel tidak dapat digunakan oleh program untuk sebagai tempat penyimpanan. Di dalam variabel tersebut, ada dapat mengakses, memodifikasi, dan menghapus nilai yang ada di dalamnya.

**3. Konsep Tipe Data dalam Python**

Tipe data adalah klasifikasi data yang menggunakan jenis nilai yang tersimpan dalam suatu variabel.Setiap bahasa pemrograman memiliki berbagai tipe data yang digunakan untuk menyimpan data dalam berbagai bentuk.Dalam Python, beberapa tipe data yang umum digunakan adalah:
 - Integer (Bilangan Bulat) : Integer adalah tipe data yang digunakan untuk menyimpan bilangan bulat.Anda dapat mendekalrasikan variabel integer dengan langsung memberikan nilai bilangan bulet ke variabel.
 - Float (Bilangan Desimal) : Float adalah tipe data yang digunakan untuk menyimpan bilangan pecahan atau desimal.Anda dapat mendeklarasikan variabel float dengan memberikan nilai desimal ke variabel atau menggunakan koma setelah bilangan bulat.
 - String (Teks) : String adalah tipe data yang digunalan untuk menyimpan teks atau huruf.Anda dapat mendeklarasikan variabel string dengan menggunakan tanda kutip tunggal (') atau ganda (").
 - Boolean (Benar/Salah) : Boolean adalah tipe data yang digunakan untuk menyimpan nilai benar(True) atau nilai salah(False).Tipe data ini sangat berguna saat menangani logika dan kondisi dalam program.

 **​4. Dynamic Typing dan Type Casting**

 Variabel dalam Python berfungsi sebagai penampung data yang dapat berubah-ubah. Python menggunakan sistem dynamic typing, di mana jenis data ditentukan secara otomatis saat nilai diberikan pada variabel.Python juga menyediakan type casting yang digunakan untuk mengubah suatu nilai dari satu tipe data ke tipe lainnya. Contohnya adalah penggunaan int(), float(), dan str(). Pemahaman mengenai type casting diperlukan agar data dapat digunakan sesuai dengan operasi yang dilakukan dan mengurangi kemungkinan terjadinya type error.

### 2.2.1 Program "Numeric_2611532012"


```python
# Buat file dengan nama Numerik_2012.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variable with numeric type value.
a_2012 = 100
print("The type of variable having value", a_2012, "is", type(a_2012))

#create a variable with float type value.
b_2012 = 10.2345
print("The type of variable having value", b_2012, "is", type (b_2012))

#create a variale with complex type value.
c_2012 = 10 + 3j
print("The type of variable having value", c_2012, "is", type(c_2012))
```

### 2.2.5 Program Driver "Konstanta_2611532012"

```python
# Buat file dengan nama konstanta_2611532012.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import final 
PI_2012: final = 3.14
print("pi: %f" % (PI_2012))
jari_2012 = float(input('Masukkan nilai jari-jari: '))
luas_2012 = PI_2012 * jari_2012
print("Luas lingkaran dengan jari - jari %.2f adalah %.2f" % (jari_2012, luas_2012))
```
